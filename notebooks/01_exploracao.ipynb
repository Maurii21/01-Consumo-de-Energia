```python
df["hora"] = df["date"].dt.hour
df["dia_semana"] = df["date"].dt.dayofweek   # 0 = segunda, 6 = domingo
df["fim_de_semana"] = df["dia_semana"] >= 5
df["mes"] = df["date"].dt.month
```

```python
consumo_por_hora = df.groupby("hora")["Appliances"].mean()
consumo_por_hora
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(x=consumo_por_hora.index, y=consumo_por_hora.values, marker="o")
plt.title("Consumo médio de aparelhos por hora do dia")
plt.xlabel("Hora do dia")
plt.ylabel("Consumo médio (Wh)")
plt.xticks(range(0, 24))
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("../images/consumo_por_hora.png", dpi=150)
plt.show()
```

### O que o gráfico mostra
- De madrugada (0h às 6h) o consumo é o mais baixo do dia, cerca de 50 Wh. Nesse horário só ficam ligados aparelhos básicos, como a geladeira.
- O consumo começa a subir por volta das 7h, quando a casa acorda e os aparelhos começam a ser usados.
- Há um primeiro pico perto das 11h (cerca de 130 Wh), seguido de uma queda no início da tarde.
- O maior pico do dia acontece às 18h, com cerca de 190 Wh. Isso é aproximadamente 4 vezes o consumo da madrugada.
- Depois das 18h o consumo cai de forma contínua até a noite.

### Conclusão
O consumo da casa segue a rotina dos moradores: é baixo quando todos dormem e alto no fim da tarde, quando provavelmente estão em casa usando vários aparelhos.

```python
df["tipo_dia"] = df["fim_de_semana"].map({False: "Dia útil", True: "Fim de semana"})

plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x="hora", y="Appliances", hue="tipo_dia",
             estimator="mean", errorbar=None)
plt.title("Consumo por hora: dias úteis vs. fim de semana")
plt.xlabel("Hora do dia")
plt.ylabel("Consumo médio (Wh)")
plt.legend(title="Tipo de dia")
plt.tight_layout()
plt.savefig("../images/dia_util_vs_fds.png", dpi=150)
plt.show()
```

### O que o gráfico mostra
- De madrugada, os dois tipos de dia são quase iguais.
- Pela manhã, o fim de semana tem consumo maior. Perto das 11h ele chega a cerca de 160 Wh, contra cerca de 120 Wh nos dias úteis.
- À noite, o pico das 18h é um pouco maior nos dias úteis (cerca de 195 Wh) do que no fim de semana (cerca de 178 Wh).

### Conclusão
Nos dias úteis, o consumo se concentra no fim da tarde. No fim de semana, ele fica mais espalhado, com um pico forte também pela manhã. Isso sugere que as pessoas ficam mais tempo em casa durante o dia aos sábados e domingos.

Observação: os dados são de uma única casa, então esse padrão vale para este conjunto de dados e não necessariamente para outras casas.

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["Appliances"])
plt.title("Distribuição do consumo de aparelhos")
plt.xlabel("Consumo (Wh)")
plt.show()
```

### O que o gráfico mostra
- A maior parte do tempo o consumo fica baixo: a caixa vai de aproximadamente 50 a 100 Wh, e a linha no meio (a mediana) está perto de 60 Wh.
- Existem muitos pontos isolados à direita, chegando a mais de 1000 Wh. Eles são os chamados outliers, ou seja, momentos de consumo muito acima do normal.

### O que decidi fazer com os outliers
Decidi **não remover** esses valores. Eles provavelmente são reais, como momentos em que vários aparelhos de alto consumo foram ligados ao mesmo tempo. Apagá-los faria o modelo ignorar justamente os picos de consumo, que são importantes.

### Conclusão
O consumo costuma ser baixo, com picos ocasionais bem altos. Por causa desses picos, a média fica maior do que o valor típico (a mediana).

```python
colunas = ["Appliances", "T_out", "RH_out", "T1", "RH_1", "hora"]
correlacao = df[colunas].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlacao, annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.title("Correlação entre consumo e variáveis ambientais")
plt.tight_layout()
plt.savefig("../images/correlacao.png", dpi=150)
plt.show()
```

### Como ler o mapa
Os números vão de -1 a +1. Quanto mais perto de 0, mais fraca é a relação em linha reta entre duas variáveis.

### O que o mapa mostra
- O consumo (Appliances) tem relação fraca com todas as variáveis. A maior é com a hora do dia (0,22).
- Com a temperatura externa (0,10), a temperatura da sala (0,06) e a umidade da sala (0,09), a relação é muito fraca.
- Com a umidade externa a relação é de -0,15: quando a umidade externa sobe, o consumo tende a cair um pouco, mas é fraco.
- Entre as próprias variáveis do clima há relações mais fortes. A temperatura externa e a temperatura da sala sobem juntas (0,68), e quando a temperatura externa sobe, a umidade externa tende a cair (-0,57).

### Conclusão
Temperatura e umidade quase não explicam o consumo. O horário é a variável que mais se relaciona com ele.

Cuidado: relação fraca em linha reta não quer dizer que não existe relação nenhuma. O consumo por hora forma "montanhas", e a correlação não consegue captar bem esse formato. Por isso, nos próximos passos vamos testar modelos que lidam melhor com isso.

## Conclusão geral

1. **O horário é o fator mais importante.** O consumo é mais baixo de madrugada (cerca de 50 Wh) e atinge o pico às 18h (cerca de 190 Wh).
2. **Dia útil e fim de semana são diferentes.** No fim de semana há um pico forte pela manhã; nos dias úteis o consumo se concentra no fim da tarde.
3. **O clima tem pouca influência.** A temperatura e a umidade têm correlação fraca com o consumo.
4. **Há picos de consumo muito altos.** Foram mantidos na análise porque parecem ser eventos reais.

### Limitações
- Os dados são de uma única casa, por cerca de 4,5 meses.
- O consumo depende do comportamento das pessoas, que não aparece nos dados.
